### BLOQUE B: cadenas de texto

El procesamiento del lenguaje natural (**NLP**, *Natural Language Processing*) trabaja con texto, así que conviene conocer bien los métodos de las cadenas (`str`).

```python
titular = "   ÚLTIMA HORA: la RAE acepta la palabra 'haiga'   "

print(titular.lower())    # todo en minúsculas
print(titular.strip())    # sin espacios al principio ni al final
print(len(titular))       # número de caracteres
```

```python
# Dividir el texto en palabras (tokens)
palabras = titular.strip().split()
print(palabras)
print(f"El titular tiene {len(palabras)} palabras")
```

```python
# ¿Aparece la palabra "RAE" en el titular?
print("RAE" in titular)

# Sustituir una palabra por otra
print(titular.replace("haiga", "haya"))
```

### BLOQUE D: bucles, condiciones y funciones

Las **funciones** nos permiten reutilizar código. Vamos a escribir una que calcule qué proporción de las letras de un texto está en mayúsculas. ¿Por qué podría ser útil para detectar noticias falsas?

```python
def proporcion_mayusculas(texto):
    """Devuelve la proporción de letras en mayúscula de un texto (entre 0 y 1)."""
    letras = [c for c in texto if c.isalpha()]
    if len(letras) == 0:
        return 0
    mayusculas = [c for c in letras if c.isupper()]
    return len(mayusculas) / len(letras)


proporcion_mayusculas("HOLA mundo")
```

```python
titulares = [
    "ÚLTIMA HORA: DESCUBREN QUE LOS GATOS ENTIENDEN EL ESPAÑOL",
    "El Gobierno aprueba los presupuestos de 2027",
    "INCREÍBLE: la NASA confirma que hay agua en Marte",
]

# Recorremos los titulares y marcamos como sospechosos los que tienen más de un 30 % de mayúsculas
for t in titulares:
    p = proporcion_mayusculas(t)
    if p > 0.3:
        print(f"¡Sospechoso! ({p:.0%}): {t}")
    else:
        print(f"Normal      ({p:.0%}): {t}")
```

Las **list comprehensions** permiten crear listas en una sola línea:

```python
# Lista con el número de caracteres de cada titular
longitudes = [len(t) for t in titulares]
longitudes
```

### BLOQUE E: primeros pasos con `pandas`

**Pandas** es la librería de Python más popular para manipular y analizar datos tabulares. Su estructura básica es el **DataFrame**: una tabla donde cada fila es una observación y cada columna una variable.

Guía: https://pandas.pydata.org/docs/getting_started/index.html

```python
import pandas as pd
```

```python
# Creamos un DataFrame a partir de un diccionario de listas
df = pd.DataFrame({
    "titular": titulares,
    "tema": ["Ciencia", "Política", "Ciencia"],
    "es_falsa": [1, 0, 0],
})
df
```

```python
# Aplicamos nuestra función a cada titular y guardamos el resultado en una nueva columna
df["mayusculas"] = df["titular"].apply(proporcion_mayusculas)
df
```

```python
# Filtramos solo las noticias falsas
df[df["es_falsa"] == 1]
```

```python
# Proporción media de mayúsculas según si la noticia es falsa o no
df.groupby("es_falsa")["mayusculas"].mean()
```

### Reto final

Escribe una función `contar_palabras` que devuelva el número de palabras de un texto y úsala para crear la columna `n_palabras`.

```python
def contar_palabras(texto):
    return len(texto.split())


df["n_palabras"] = df["titular"].apply(contar_palabras)
df
```
